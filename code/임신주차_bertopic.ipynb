```python
import pandas as pd

# 주차 범위 설정 딕셔너리
title = {
    1: [0, 7], 2: [8, 11], 3: [12, 15], 4: [16, 19], 5: [20, 23],
    6: [24, 27], 7: [28, 31], 8: [32, 35], 9: [36, 39], 10: [40, 40]
}

df_list = []

for num in range(1, 11):
    start_wk, end_wk = title[num]
    file_path = f"./momsholic_{start_wk}_{end_wk}.csv"

    try:
        # 데이터 로드
        temp_df = pd.read_csv(file_path)
        temp_df = temp_df.iloc[32:, :]

        # 주차 식별을 위한 week 열 생성 (해당 범위의 시작 주차로 표기하거나 범위를 텍스트로 저장)
        # BERTopic의 topics_over_time에 넘길 timestamps는 숫자형이어야 하므로 시작 주차(start_wk)를 할당
        temp_df['week'] = start_wk

        df_list.append(temp_df)
        print(f"✅ Success: {file_path} (Rows: {len(temp_df)})")

    except FileNotFoundError:
        print(f"❌ Error: {file_path} 파일을 찾을 수 없습니다.")

# 전체 데이터 병합
df = pd.concat(df_list, ignore_index=True)

# BERTopic 입력용 리스트 생성
docs = df['Content'].astype(str).tolist()
timestamps = df['week'].tolist()

print(f"\nTotal Data Rows: {len(df)}")
```

```python
for num in range(10, 11):
    start_wk, end_wk = title[num]
    file_path = f"./momsholic_{start_wk}_{end_wk}.csv"

    try:
        # 1. 데이터 로드 및 결측치 즉시 제거
        df_sub = pd.read_csv(file_path).dropna(subset=['Content'])
        df_sub = df_sub.iloc[32:, :].reset_index(drop=True)

        # 2. 방어적 전처리 루프
        docs = []
        for doc in df_sub['Content']:
            try:
                # 입력을 강제로 string으로 변환 후 전처리
                cleaned = clean_text(str(doc))
                if isinstance(cleaned, str) and len(cleaned.strip()) >= 2:
                    docs.append(cleaned.strip())
            except:
                continue

        if not docs:
            print(f"⚠️ {start_wk}주차: 분석할 유효 데이터가 없습니다.")
            continue

        print(f"\n🚀 [Processing] {start_wk}-{end_wk}주차 ({len(docs)}건)")

        # 3. [핵심] CountVectorizer 내부 에러 방지 설정
        # lowercase=False를 주어 추가 변환 과정에서의 타입 오류 방지
        # 이미 clean_text에서 정제되었으므로 안전함
        vectorizer_model = CountVectorizer(
            stop_words=stopwords,
            min_df=5,
            token_pattern=r"(?u)\b\w\w+\b" # 한 글자 단어 제외 정규식
        )

        umap_model = UMAP(n_neighbors=15, n_components=5, min_dist=0.0, metric='cosine', random_state=42)
        hdbscan_model = HDBSCAN(min_cluster_size=10, metric='euclidean', cluster_selection_method='eom', prediction_data=True)

        topic_model = BERTopic(
            embedding_model=embedding_model,
            umap_model=umap_model,
            hdbscan_model=hdbscan_model,
            vectorizer_model=vectorizer_model,
            verbose=False
        )

        # 4. 학습 시 리스트 타입 재확인
        # 만에 하나 섞여 있을 float를 위해 다시 한번 str() 랩핑
        docs = [str(d) for d in docs]

        topics, _ = topic_model.fit_transform(docs)

        if len(set(topics)) > 1:
            fig = topic_model.visualize_barchart(
                top_n_topics=10,
                n_words=10,
                height=600,
                title=f"Topic Analysis: {start_wk}-{end_wk} Weeks"
            )
            fig.show()
        else:
            print(f"ℹ️ {start_wk}주차: 추출된 유의미한 토픽이 없습니다.")

    except Exception as e:
        print(f"❌ {start_wk}주차 최종 처리 실패: {e}")
        # 구체적으로 어디서 에러가 나는지 확인하기 위해 docs의 첫 5개 값 출력
        if 'docs' in locals() and docs:
            print(f"Docs Sample: {docs[:5]}")

    finally:
        if 'topic_model' in locals():
            del topic_model
        torch.cuda.empty_cache()
```
